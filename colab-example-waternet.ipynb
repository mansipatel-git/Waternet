## Init network

```python
import torch
import cv2
import matplotlib.pyplot as plt
%matplotlib inline
```

```python
# Make sure you are connected to a GPU instance
torch.cuda.is_available()
```

```python
# Load from torchhub
preprocess, postprocess, model = torch.hub.load('tnwei/waternet', 'waternet')
model.eval();
```

```python
# Resize image
rgb_im = cv2.resize(rgb_im, (720, 480))
```

```python
# Inference -> return numpy array (1, 3, H, W)
rgb_ten, wb_ten, he_ten, gc_ten = preprocess(rgb_im)
out_ten = model(rgb_ten, wb_ten, he_ten, gc_ten)
out_im = postprocess(out_ten)
```

```python
fig, ax = plt.subplots(ncols=2, figsize=(14, 5))
ax[0].imshow(rgb_im)
ax[0].axis("off")
ax[0].set_title("Original image")

ax[1].imshow(out_im[0])
ax[1].axis("off")
ax[1].set_title("WaterNet output")

plt.show()
```

```python
# ============================================================
# QUALITY METRICS
# ============================================================

import numpy as np
import cv2
from skimage.metrics import mean_squared_error
from skimage.metrics import peak_signal_noise_ratio
from skimage.metrics import structural_similarity


# ------------------------------------------------------------
# Convert images to uint8
# ------------------------------------------------------------

original = np.clip(rgb_im, 0, 255).astype(np.uint8)
enhanced = np.clip(out_im, 0, 255).astype(np.uint8)


# ============================================================
# 1. MSE
# ============================================================

mse_value = mean_squared_error(original, enhanced)


# ============================================================
# 2. PSNR
# ============================================================

psnr_value = peak_signal_noise_ratio(
    original,
    enhanced,
    data_range=255
)


# ============================================================
# 3. SSIM
# ============================================================

ssim_value = structural_similarity(
    original,
    enhanced,
    channel_axis=2,
    data_range=255
)


# ============================================================
# PRINT RESULTS
# ============================================================

print("======================================")
print("      WATER-NET QUALITY METRICS")
print("======================================")

print(f"MSE  : {mse_value:.4f}")
print(f"PSNR : {psnr_value:.4f} dB")
print(f"SSIM : {ssim_value:.4f}")
```
